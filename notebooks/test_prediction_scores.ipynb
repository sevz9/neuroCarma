```python
import plotly.express as px
import pandas as pd
import numpy as np

import plotly.graph_objects as go
```

```python
tables = [pd.read_csv("./pulsar_data.csv"), pd.read_csv("./pulsar_data_2.csv")]

EPOCHS_COL = "# Epochs"
BATCH_COL = "# Nodes in a batch"
MAX_NUM_NEIGHBORS_COL = "Max neighbours for a node"

CORR_COL = "Correlation w/ original model"
pulsar_dataset: pd.DataFrame = pd.concat(tables)[["num_epochs", "batch_size", "max_num_neighbors", "pearson_correlation"]].dropna().drop_duplicates()
pulsar_dataset.loc[pulsar_dataset["max_num_neighbors"] == -1, "max_num_neighbors"] = -100
pulsar_dataset["max_num_neighbors"] = pulsar_dataset["max_num_neighbors"].astype("category")

pulsar_dataset.rename(mapper={
    "num_epochs": EPOCHS_COL,
    "batch_size": BATCH_COL,
    "max_num_neighbors": MAX_NUM_NEIGHBORS_COL,
    "pearson_correlation": CORR_COL
},
                      inplace=True,
                      axis="columns"
                      )
pulsar_dataset.head()
```

```python
pulsar_dataset.shape
```

```python

params = dict(zip("xyz", [EPOCHS_COL, BATCH_COL, MAX_NUM_NEIGHBORS_COL]))
fig = px.scatter_3d(pulsar_dataset, **params, color=CORR_COL, log_y=True)

fig.update_layout(height=800, width=1200)
fig.show()
```

```python
fig.write_html("./pulsar_batch_exp.html")
```

```python

from itertools import combinations

datasets = {}
for col_1, col_2 in combinations([MAX_NUM_NEIGHBORS_COL, BATCH_COL, EPOCHS_COL], 2):
    datasets[(col_1, col_2)] = datasets[(col_2, col_1)] = pulsar_dataset.groupby(by=[col_1, col_2])[CORR_COL].mean().reset_index()
```

```python
pulsar_dataset.groupby(by=[col_1, col_2])[CORR_COL].mean().reset_index()
```

```python
pairs_params = [
    (BATCH_COL, MAX_NUM_NEIGHBORS_COL),
    (EPOCHS_COL, BATCH_COL),
    (EPOCHS_COL, MAX_NUM_NEIGHBORS_COL)
]
```

```python
for pair in pairs_params:
    x, y = pair
    
    df = datasets[(x, y)]
    px.scatter(df, x=x, y=y, color=CORR_COL)
```
